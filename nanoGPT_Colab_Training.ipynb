### Downloading your trained checkpoint
Once training finishes, the best model is saved in `checkpoints_v2/`. You can download it directly.

```
from google.colab import files
import os

# Assuming it trains for 3000 iterations
checkpoint_path = 'checkpoints_v2/ckpt_03000.pt'
if os.path.exists(checkpoint_path):
    files.download(checkpoint_path)
else:
    print('Checkpoint not found. Make sure training completed successfully.')
```
